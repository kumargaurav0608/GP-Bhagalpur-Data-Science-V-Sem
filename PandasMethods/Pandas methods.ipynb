```python
# 4. Data Cleaning & Transformation - Handle missing data, duplicates, and structurally alter column values.
#     dataset.isnull() / dataset.isna() - Check for missing or NaN values across the dataset 
dataset.isnull()
```

```python
# for counting columns has how many null values.
dataset.isna().sum()
```

```python
#create a copy of your datset 
clean_dataset = dataset.copy()
```

```python
# check  duplicate rows from the DataFrame.
clean_dataset.duplicated().sum()
```

```python
#remove duplicates 
clean_dataset.drop_duplicates()
```

```python
# Remove rows or columns that contain missing values
# clean_dataset.dropna(axis=0)  # Remove rows
# clean_dataset.dropna(axis=1)  # Remove columns
clean_dataset.dropna()
```

```python
#want to modify original dataset
clean_dataset.dropna(inplace=True)
```

```python
clean_dataset.shape
```

```python
clean_dataset.dropna(axis=0)
```

```python
# df.fillna(value)
# fillna() is used to replace missing (NaN) values.
clean_dataset.fillna(0)
```

```python
# Replace column  using a specific values
clean_dataset["UG_fee"] = clean_dataset["UG_fee"].fillna(clean_dataset["UG_fee"].mean())
```

```python
# apply() is used to apply a function to values in a DataFrame or Series.
# ncrease all fees by 10%
clean_dataset["UG_fee"] = clean_dataset["UG_fee"].apply(lambda x: x * 1.10)
```

```python
# 5. Aggregation & GroupingSummarize data metrics by segmenting your columns
# dataset.groupby() - Groups rows based on one or more columns so you can perform calculations on each group
clean_dataset.groupby("College_Name")["UG_fee"].mean()
```

```python
# dataset.agg()	Applies one or multiple aggregation functions, such as sum, mean, min, or max
clean_dataset.groupby("College_Name")["UG_fee"].agg(["mean", "max", "min"])
```

```python
# dataset.transform() 	Performs calculations within groups while keeping the same number of rows as the original DataFrame
clean_dataset.groupby("College_Name")["UG_fee"].transform("mean")
```

```python
# dataset.value_counts(): Count the occurrences of unique values within a single column.
clean_dataset["College_Name"].value_counts()
```

```python
# 6. Reshaping & Combining
# These pandas functions are used to combine datasets or change the structure/orientation of a DataFrame.
    
# ps.merge()	- Joins two DataFrames using common columns/keys - 	Join tables (inner, Outer , Right join, Left Join)
#     ps.merge(dataset1, dataset2, on="ID", how="inner")
    
# ps.concat()	- Combines DataFrames vertically or horizontally	- Stack tables
#     ps.concat([dataset1, dataset2], axis=0) // for row
#     ps.concat([dataset1, dataset2], axis=1) // for column

# dataset.pivot_table()	- Creates a summarized table using rows, columns, and aggregation - 	Summarize & reshape
clean_dataset.pivot_table( values="UG_fee",index="College_Name", columns="State",aggfunc="sum")

```
