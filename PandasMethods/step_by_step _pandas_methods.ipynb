```python
# 4. Check and handle missing values
# Find missing values:
df.isnull().sum()
```

```python
#change data type of the column 
df["Upload Year"] = pd.to_numeric(df["Upload Year"], errors="coerce")

df["Upload Year"] = df["Upload Year"].fillna(
    df["Upload Year"].mean().round()
).astype(int)
```

```python
# Fill missing values:
df["Upload Year"] = df["Upload Year"].fillna(df["Upload Year"].mean())
```

```python
#for number fill with 0 
numeric_cols = df.select_dtypes(include='number').columns
df[numeric_cols] = df[numeric_cols].fillna(0)
#for string fill with unknown
string_cols = df.select_dtypes(include='str').columns
df[string_cols] = df[string_cols].fillna("Unknown")
```

```python
# 5. Check and remove duplicates
# Find duplicates:
df.duplicated().sum()
```

```python
# Remove duplicates:
df = df.drop_duplicates()
```

```python
df.dropna(inplace=True) #for row deleting missing values
```

```python
# df.head(5)
```

```python
# 6. Check and correct data types
# Check data types:
df.dtypes
```

```python
df = df.rename(columns={
    "Year of Establishment\n": "Year_of_Establishment",
    "The Institute Added In Survey Year ": "Inst_added_in_survey_year",
    "Upload Year": "Upload_Year",
    "Specialised in": "Specialised_in",
    "University Type": "University_Type",
    "University Name": "University_Name",
    "College Name": "College_Name",
    "College Type": "College_Type"
})
```

```python
#clean all automatically
df.columns = (
    df.columns
    .str.strip()
    .str.replace("\n", "", regex=False)
    .str.replace(" ", "_")
)
```
